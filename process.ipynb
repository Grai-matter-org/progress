```python
import pandas as pd
ICD_ALL_COLUMNS=UICD
KEEP_COLUMNS=['patient_id', 'encounter_id', 'n_icd_prefix3',
       'encounter_setting', 'encounter_medical_service','encounter_duration_days',
        'PATIENT_SEX', 'PATIENT_RACE_ETHNICITY',
       'age']
def expand_icd_prefix3(
    df: pd.DataFrame,
    KEEP_COLUMNS: list,
    ICD_ALL_COLUMNS: list,
    icd_col: str = "icd_prefix3_unique_str",
    use_sparse: bool = True
) -> pd.DataFrame:
    # Keep the requested columns (they’ll be left-most in the output)
    base = df[KEEP_COLUMNS].copy()

    # Handle missing/empty strings safely
    icd_str = df[icd_col].fillna("").astype(str).str.strip()

    # Fast one-hot over comma-separated prefixes → 0/1 per row
    dummies = icd_str.str.get_dummies(sep=",")

    # Ensure we only keep the predefined ICD columns, preserving order
    icd_mat = dummies.reindex(columns=ICD_ALL_COLUMNS, fill_value=0)

    # Make sure values are integers (and optionally sparse to save RAM)
    if use_sparse:
        icd_mat = icd_mat.astype(pd.SparseDtype("int", 0))
    else:
        icd_mat = icd_mat.astype("int16")

    # Concatenate: [your kept columns | ICD columns]
    out = pd.concat([base, icd_mat], axis=1)
    return out
```

```python
import pandas as pd
from typing import Dict, List, Tuple, Optional, Union

def _is_all_numeric_like(series: pd.Series) -> bool:
    """True if non-null values are numeric or numeric strings."""
    s = series.dropna().astype(str).str.strip()
    if s.empty:
        return True
    # numeric if everything parses with to_numeric (ignore errors -> NaN, then check)
    parsed = pd.to_numeric(s, errors="coerce")
    return not parsed.isna().any()

def encode_columns_to_ints(
    df: pd.DataFrame,
    columns: List[str],
    return_format: str = "dict",       # "dict" | "list"
    existing_maps: Optional[Dict[str, Dict[str, int]]] = None,
    include_unknown_token: bool = True,
    unknown_code: int = -1,
    sort_labels: bool = True,
    keep_original: bool = False,
    original_suffix: str = "_orig"
) -> Tuple[pd.DataFrame, Union[Dict[str, Dict[str, int]], List[Dict[str, int]]]]:
    """
    For each column in `columns`:
      • If values are numeric-but-strings -> cast to integer dtype (no mapping needed; returns {} for that col).
      • Else build a stable categorical→int mapping and transform the column to int.
        - If `existing_maps` is provided for a column, use it and map unseen to `unknown_code` when enabled.
        - Otherwise fit a new mapping from the data in that column.

    Parameters
    ----------
    df : DataFrame
        Input data.
    columns : list of str
        Columns to process.
    return_format : {"dict","list"}
        Return per-column mappings as a dict-of-dicts keyed by column name, or
        as a list of dicts in the same order as `columns`.
    existing_maps : dict or None
        Optional pre-fit mappings: {col_name: {label: code}}.
    include_unknown_token : bool
        If True and a mapping is used, unseen labels get `unknown_code`.
        If False, unseen labels raise an error.
    unknown_code : int
        Code to use for unseen labels.
    sort_labels : bool
        If True, when fitting a new mapping, sort unique labels for deterministic codes.
        If False, preserve first-seen order.
    keep_original : bool
        If True, keep a copy of the original column as `<col>_orig`.
    original_suffix : str
        Suffix for the original column copy if `keep_original=True`.

    Returns
    -------
    transformed_df : DataFrame
        DataFrame with specified columns transformed to integer dtype.
    mappings : dict or list
        Per-column mapping(s). For columns that were simply numeric-casted,
        an empty dict is returned for that column.
    """
    if return_format not in {"dict", "list"}:
        raise ValueError("return_format must be 'dict' or 'list'")

    df_out = df.copy()
    col_maps: Dict[str, Dict[str, int]] = {}

    existing_maps = existing_maps or {}

    for col in columns:
        if col not in df_out.columns:
            raise KeyError(f"Column '{col}' not found in DataFrame.")

        if keep_original:
            df_out[f"{col}{original_suffix}"] = df_out[col]

        # Case 1: numeric-like → cast to integer
        if _is_all_numeric_like(df_out[col]):
            # coerce to integer (preserve NaN as pd.NA then to Int64)
            coerced = pd.to_numeric(df_out[col], errors="coerce")
            # If you must ensure *plain* int for non-null columns, you can fillna then cast:
            # df_out[col] = coerced.fillna(0).astype("int64")
            # But better to use nullable integer to preserve missingness:
            df_out[col] = coerced.astype("Int64")
            col_maps[col] = {}  # No categorical mapping needed
            continue

        # Case 2: categorical strings → map to ints
        # Use existing map if provided; otherwise fit
        fit_map = existing_maps.get(col)
        if fit_map is None:
            # Fit a new mapping
            vals = df_out[col].astype("string")  # robust to NaN
            uniq = pd.Index(vals.dropna().unique())
            if sort_labels:
                uniq = pd.Index(sorted(uniq))
            fit_map = {label: i for i, label in enumerate(uniq)}
            col_maps[col] = fit_map
        else:
            # Use provided mapping as-is
            col_maps[col] = dict(fit_map)  # copy

        # Apply mapping with unknown handling
        def _map_value(v):
            if pd.isna(v):
                return pd.NA
            v = str(v)
            if v in fit_map:
                return fit_map[v]
            if include_unknown_token:
                return unknown_code
            raise KeyError(f"Unseen label '{v}' in column '{col}' and include_unknown_token=False.")

        df_out[col] = df_out[col].map(_map_value).astype("Int64")

    if return_format == "dict":
        return df_out, col_maps
    else:
        # list aligned to `columns`
        return df_out, [col_maps.get(c, {}) for c in columns]
```

```python

import pandas as pd
import numpy as np

def add_prior_encounter_sums_for_historical_fast(
    ef__: pd.DataFrame,
    df: pd.DataFrame,
    HISTORICAL_COLUMNS: list,
    patient_col: str = "patient_id",
    enc_id_col: str = "encounter_id",
    start_col: str = "encounter_start_dt",
    end_col: str = "encounter_end_dt",
    gap_fallback_days: int = 1000,
    gap_col_name: str = "days_since_last_encounter",
    dur_mean_col: str = "prior_enc_length_mean_days",
    dur_std_col: str = "prior_enc_length_std_days",
) -> pd.DataFrame:
    ef_aug = ef__.copy()
    src = df.copy()

    # ensure encounter_id is a column
    had_index = (ef_aug.index.name == enc_id_col)
    if had_index:
        ef_aug = ef_aug.reset_index()
    elif enc_id_col not in ef_aug.columns:
        raise KeyError(f"'{enc_id_col}' not found as index or column in ef__")

    # restore patient/start/end if missing in ef__
    need = [patient_col, start_col, end_col]
    missing = [c for c in need if c not in ef_aug.columns]
    if missing:
        if not set(need).issubset(src.columns):
            absent = [c for c in need if c not in src.columns]
            raise KeyError(f"ef__ missing {missing} and df lacks {absent} to restore.")
        lut = src[[enc_id_col] + need].drop_duplicates(enc_id_col)
        ef_aug = ef_aug.merge(lut, on=enc_id_col, how="left")

    # datetimes
    ef_aug[start_col] = pd.to_datetime(ef_aug[start_col], errors="coerce")
    ef_aug[end_col]   = pd.to_datetime(ef_aug[end_col], errors="coerce")
    src[start_col]    = pd.to_datetime(src[start_col], errors="coerce")
    src[end_col]      = pd.to_datetime(src[end_col], errors="coerce")

    # validate historical columns
    miss_hist = [c for c in HISTORICAL_COLUMNS if c not in src.columns]
    if miss_hist:
        raise KeyError(f"HISTORICAL_COLUMNS not in df: {miss_hist}")

    # result buffers
    out_cols = [f"{c}_prior_sum" for c in HISTORICAL_COLUMNS]
    result_sums = np.zeros((len(ef_aug), len(HISTORICAL_COLUMNS)), dtype=np.int64)
    result_gap  = np.full(len(ef_aug), gap_fallback_days, dtype=np.int64)
    # mean/std will be float with NA where no priors
    result_mean = np.full(len(ef_aug), np.nan, dtype=np.float64)
    result_std  = np.full(len(ef_aug), np.nan, dtype=np.float64)

    # only rows with valid patient & start are queries
    valid_left = (~ef_aug[patient_col].isna()) & (~ef_aug[start_col].isna())
    ef_aug["_row_ix"] = np.arange(len(ef_aug))

    # iterate per-patient
    for pid, left_block in ef_aug.loc[valid_left, [patient_col, start_col, "_row_ix"]].groupby(
        ef_aug.loc[valid_left, patient_col].values
    ):
        # right = all encounters for pid with valid start & end (needed for duration)
        right_block = src.loc[src[patient_col] == pid, [start_col, end_col] + HISTORICAL_COLUMNS] \
                        .dropna(subset=[end_col, start_col])
        if right_block.empty:
            continue

        # sort by end time and build cumulative sums
        right_block = right_block.sort_values(end_col, kind="mergesort")
        ends = right_block[end_col].to_numpy()

        # cumulative sums for requested historical columns
        cum = np.cumsum(
            right_block[HISTORICAL_COLUMNS].fillna(0).to_numpy(dtype=np.float64),
            axis=0
        )

        # encounter durations (days) and their prefix sums for mean/std
        # ensure non-negative durations (clip at 0)
        durs = (right_block[end_col].to_numpy() - right_block[start_col].to_numpy()) \
                    .astype("timedelta64[D]").astype(np.int64)
        durs = np.clip(durs, 0, None).astype(np.float64)
        cum_dur = np.cumsum(durs)
        cum_dur2 = np.cumsum(durs * durs)
        counts = np.arange(1, len(right_block) + 1, dtype=np.int64)

        # strictly prior: end < start
        starts = left_block[start_col].to_numpy()
        idxs = np.searchsorted(ends, starts, side='left') - 1  # last index with end < start

        # gather cumulative sums for historical columns
        prior_vals = np.zeros((len(starts), len(HISTORICAL_COLUMNS)), dtype=np.float64)
        ok = idxs >= 0
        if ok.any():
            prior_vals[ok] = cum[idxs[ok]]

        # compute gap (days since last prior end)
        if ok.any():
            last_ends = ends[idxs[ok]]
            gaps = (starts[ok] - last_ends).astype("timedelta64[D]").astype(np.int64)
            gaps = np.clip(gaps, 0, None)
        else:
            gaps = np.array([], dtype=np.int64)

        # compute mean/std of prior encounter lengths (days) using prefix sums
        # for each row with ok, we have n = counts[idx], sum = cum_dur[idx], sumsq = cum_dur2[idx]
        if ok.any():
            n = counts[idxs[ok]].astype(np.float64)
            s1 = cum_dur[idxs[ok]]
            s2 = cum_dur2[idxs[ok]]
            mean = s1 / n
            # population std by default; change to (n-1) for sample std if desired
            var = s2 / n - mean * mean
            var = np.clip(var, 0.0, None)
            std = np.sqrt(var)
        else:
            mean = np.array([], dtype=np.float64)
            std  = np.array([], dtype=np.float64)

        # place back by original row indices
        rows = left_block["_row_ix"].to_numpy()
        result_sums[rows, :] = prior_vals.astype(np.int64, copy=False)
        if ok.any():
            result_gap[rows[ok]]  = gaps
            result_mean[rows[ok]] = mean
            result_std[rows[ok]]  = std

    # attach outputs
    for j, col in enumerate(out_cols):
        ef_aug[col] = pd.array(result_sums[:, j], dtype="Int64")
    ef_aug[gap_col_name] = pd.array(result_gap, dtype="Int64")
    ef_aug[dur_mean_col] = pd.array(result_mean, dtype="Float64")
    ef_aug[dur_std_col]  = pd.array(result_std, dtype="Float64")

    # cleanup
    ef_aug = ef_aug.drop(columns=["_row_ix"])

    # restore index layout if needed
    if had_index:
        ef_aug = ef_aug.set_index(enc_id_col).sort_index()

    return ef_aug
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm
import pandas as pd
import seaborn as sns
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn import neighbors, datasets
from sklearn.model_selection import cross_val_score
from sklearn.datasets import make_blobs
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.ensemble import GradientBoostingClassifier
from scipy.spatial import ConvexHull
from tqdm import tqdm
import random
plt.style.use('ggplot')
import pickle
from sklearn import tree
from sklearn.tree import export_graphviz
from joblib import dump, load
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from scipy.interpolate import interp1d
import warnings
warnings.filterwarnings("ignore")

def getAuc(X,y,test_size=0.25,max_depth=None,n_estimators=100,
           minsplit=4,FPR=[],TPR=[],VERBOSE=False, USE_ONLY=None):
    '''
        get AUC given training data X, with target labels y
    '''
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size)
    CLASSIFIERS=[DecisionTreeClassifier(max_depth=max_depth, min_samples_split=minsplit,class_weight='balanced'),
                RandomForestClassifier(n_estimators=n_estimators,
                                       max_depth=max_depth,min_samples_split=minsplit,class_weight='balanced'),
                ExtraTreesClassifier(n_estimators=n_estimators,
                                     max_depth=max_depth,min_samples_split=minsplit,class_weight='balanced'),
                AdaBoostClassifier(n_estimators=n_estimators),
                GradientBoostingClassifier(n_estimators=n_estimators,max_depth=max_depth),
                svm.SVC(kernel='rbf',gamma='scale',class_weight='balanced',probability=True)]

    if USE_ONLY is not None:
        if isinstance(USE_ONLY, (list,)):
            CLASSIFIERS=[CLASSIFIERS[i] for i in USE_ONLY]
        if isinstance(USE_ONLY, (int,)):
            CLASSIFIERS=CLASSIFIERS[USE_ONLY]

    for clf in CLASSIFIERS:
        clf.fit(X_train,y_train)
        y_pred=clf.predict_proba(X_test)
        fpr, tpr, thresholds = metrics.roc_curve(y_test,y_pred[:,1], pos_label=1)
        auc=metrics.auc(fpr, tpr)
        if VERBOSE:
            print(clf,auc)

        FPR=np.append(FPR,fpr)
        TPR=np.append(TPR,tpr)
        plt.plot(fpr,tpr)
    points=np.array([[a[0],a[1]] for a in zip(FPR,TPR)])
    hull = ConvexHull(points)
    x=np.argsort(points[hull.vertices,:][:,0])
    auc=metrics.auc(points[hull.vertices,:][x,0],points[hull.vertices,:][x,1])
    return auc,CLASSIFIERS

def saveFIG(filename='tmp.pdf',AXIS=False):
    '''
        save fig for publication
    '''
    import pylab as plt
    plt.subplots_adjust(top = 1, bottom = 0, right = 1, left = 0, 
            hspace = 0, wspace = 0)
    plt.margins(0,0)
    if not AXIS:
        plt.gca().xaxis.set_major_locator(plt.NullLocator())
        plt.gca().yaxis.set_major_locator(plt.NullLocator())
    plt.savefig(filename,dpi=300, bbox_inches = 'tight',
                pad_inches = 0,transparent=False) 
    return
```
